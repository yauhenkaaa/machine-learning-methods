```python
sec = int(input("Введите количество секунд:"))
while (sec != 0):
    days = sec // (24 * 3600)
    rem = sec % (24 * 3600)
    hours = rem // 3600
    rem = sec % 3600
    minutes = rem // 60
    seconds = sec % 60
    print(f"{days}:{hours:02d}:{minutes:02d}:{seconds:02d}")
    sec = int(input("Введите количество секунд:"))
```

### Задание 3.
На вход программе подается натуральное число n, а затем n целых чисел. Напишите программу, которая создает из указанных чисел список их кубов. Найти сумму и произведение элементов списка. Вывести список в обратном порядке.

```python
n = int(input("Введите число n: "))

cubes = []
total_sum = 0
product = 1

for i in range(n):
    number = int(input(f"Введите число №{i + 1}: "))
    cube = number ** 3

    cubes.append(cube)
    total_sum += cube
    product *= cube

cubes.reverse()

print(f"Сумма: {total_sum}")
print(f"Произведение: {product}")
print(f"Список в обратном порядке: {cubes}")
```

### Задание 5. 
Реализуйте программу «Магазин игрушек», которая будет включать в себя шесть пунктов меню.  
У вас есть словарь, где ключ – название игрушки. Значение – список, который содержит состав игрушки, цену и кол-во (шт), которое есть в магазине.  
1. Просмотр описания: название – описание.
2. Просмотр цены: название – цена.
3. Просмотр количества: название – количество.
4. Всю информацию.
5. Покупка. В пункте «Покупка» необходимо совершить покупку, с клавиатуры вводите название игрушки и его кол-во, n – выход из программы. Посчитать цену выбранных товаров и сколько товаров осталось в изначальном списке. 
6. До свидания

```python
toys = {
    'Мишка':   ['плюш, синтепон', 25.5, 10],
    'Машинка': ['пластик, металл', 12.0, 15],
    'Кукла':   ['пластик, ткань', 30.0, 7],
    'Конструктор': ['пластик', 45.9, 5],
}

while True:
    print('\n=== Магазин игрушек ===')
    print('1. Просмотр описания')
    print('2. Просмотр цены')
    print('3. Просмотр количества')
    print('4. Вся информация')
    print('5. Покупка')
    print('6. До свидания')
    choice = input('Выберите пункт (1-6): ').strip()

    if choice == '1':
        for name, data in toys.items():
            print(f'{name} - {data[0]}')

    elif choice == '2':
        for name, data in toys.items():
            print(f'{name} - {data[1]} руб.')

    elif choice == '3':
        for name, data in toys.items():
            print(f'{name} - {data[2]} шт.')

    elif choice == '4':
        for name, data in toys.items():
            print(f'{name}: состав - {data[0]}; цена - {data[1]} руб.; количество - {data[2]} шт.')

    elif choice == '5':
        total = 0
        bought = {}
        while True:
            name = input('Название игрушки (n - завершить покупку): ').strip().capitalize()
            if name.lower() == 'n':
                break
            if name not in toys:
                print('Такой игрушки нет в магазине.')
                continue

            qty_text = input('Количество: ').strip()
            if not qty_text.isdigit() or int(qty_text) <= 0:
                print('Введите целое положительное число.')
                continue
            qty = int(qty_text)

            if qty > toys[name][2]:
                print(f'В наличии только {toys[name][2]} шт.')
                continue

            toys[name][2] -= qty
            total += qty * toys[name][1]
            bought[name] = bought.get(name, 0) + qty
            print(f'Добавлено: {name} x {qty}')

        print('\n--- Чек ---')
        if bought:
            for name, qty in bought.items():
                print(f'{name}: {qty} шт. x {toys[name][1]} = {qty * toys[name][1]:.2f} руб.')
            print(f'Итого к оплате: {total:.2f} руб.')
        else:
            print('Вы ничего не купили.')

        print('\nОстаток в магазине:')
        for name, data in toys.items():
            print(f'{name} - {data[2]} шт.')

    elif choice == '6':
        print('До свидания!')
        break

    else:
        print('Неверный пункт меню, попробуйте снова.')
```

### Задание №2
Напишите функцию, которая будет принимать один аргумент. 
- Если в функцию передаётся список, найти сумму всех элементов после последнего положительного. Удалить все нулевые элементы из списка.  
- Если словарь, вывести элемент с минимальным значением.  
- Число – Вывести в обратном порядке. 
- Строка – определить количество слов.  

Сделать проверку со всеми этими случаями. 

```python
def check_argument(x):
    if type(x) is list:
        last_pos = -1
        for i in range(len(x)):
            if x[i] > 0:
                last_pos = i
                
        if last_pos == -1:
            print('Положительных элементов нет')
        else:
            print('Сумма после последнего положительного:', sum(x[last_pos + 1:]))

        x[:] = [y for y in x if y != 0] 
        print('Список без нулей:', x)
        return x

    elif type(x) is dict:
        if not x:
            print('Словарь пуст')
            return None
        key = min(x, key=x.get)
        print('Минимальный элемент:', key, ':', x[key])
        return key, x[key]

    elif type(x) is int:
        text = str(x)
        if text.startswith('-'):
            rev_text = '-' + text[:0:-1]
        else:
            rev_text = text[::-1]
        result = int(rev_text)
        print (f"Развернутое число: {result}")
        return result

    elif type(x) is str:
        count = len(x.split())
        print('Количество слов:', count)
        return count



print('--- список ---')
check_argument([0, -1, 3, -5, 0])

print('--- словарь ---')
check_argument({'a': 500, 'b': 20, 'c': 560})

print('--- число ---')
check_argument(12345)
check_argument(-1200)

print('--- строка ---')
check_argument('Привет, как дела у тебя?')
```

### Задание №3

```python
def check_matrix(x):
    if all(any(el > 0 for el in row) for row in x):
        x = [[-el for el in row] for row in x]
        print('Знаки изменены:')
    else:
        print('Есть строка без положительных элементов, матрица не изменена:')

    for row in x:
        print(row)
    
    return x

matrix = [
    [1, -2, 3],
    [-4, 5, -6],
    [-7, -8, 9],
]

check_matrix(matrix)
```

### Задание №4

Напишите программу, демонстрирующую работу try\except\finally.

```python
def divide_numbers():
    try:
        a = int(input('Введите делимое: '))
        b = int(input('Введите делитель: '))
        result = a / b
    except ValueError:
        print('Ошибка: нужно вводить целые числа.')
    except ZeroDivisionError:
        print('Ошибка: на ноль делить нельзя.')
    else:
        print(f'Результат: {result}')
    finally:
        print('Блок finally выполнен всегда.\n')


divide_numbers()   
divide_numbers()   
divide_numbers() 
```

### Задание №1: 
Создать класс String (строка). Разработать в классе методы для работы со строками (не менее 5).

```python
class String:

    def __init__(self, text=''):
        self.text = text

    def word_count(self):
        return len(self.text.split())

    def reverse(self):
        return self.text[::-1]

    def is_palindrome(self):
        s = ''.join(ch.lower() for ch in self.text if ch.isalnum())
        return s == s[::-1]

    def count_char(self, char):
        return self.text.count(char)

    def vowels_count(self):
        return sum(1 for ch in self.text.lower() if ch in 'aeiouyаеёиоуыэюя')

    def longest_word(self):
        words = self.text.split()
        return max(words, key=len) if words else ''

    def __str__(self):
        return self.text


print('=== Задание 1 ===')
s = String('А роза упала на лапу Азора')
print('Слов:', s.word_count())
print('Наоборот:', s.reverse())
print('Палиндром:', s.is_palindrome())
print("Символов 'а':", s.count_char('а'))
print('Гласных:', s.vowels_count())
print('Самое длинное слово:', s.longest_word())
```

### Задание №2

Создать классы Alphabet, EnglishAlphabet

```python
class Alphabet:
    def __init__(self, lang, letters):
        self.lang = lang
        self.letters = list(letters)

    def print(self):
        print(' '.join(self.letters))

    def letters_num(self):
        return len(self.letters)


class EngAlphabet(Alphabet):
    __letters_num = 26                      # приватное статическое свойство

    def __init__(self):
        super().__init__('En', "ABCDEFGHIJKLMNOPQRSTUVWXYZ")

    def is_en_letter(self, letter):
        return len(letter) == 1 and letter.upper() in self.letters

    def letters_num(self):
        return EngAlphabet.__letters_num

    @staticmethod
    def example():
        return 'The quick brown fox jumps over the lazy dog.'

print('\n=== Задание 2 ===')
eng = EngAlphabet()
eng.print()
print('Букв:', eng.letters_num())
print("is_en_letter('F'):", eng.is_en_letter('F'))
print("is_en_letter('Ж'):", eng.is_en_letter('Ж'))
print('Пример:', EngAlphabet.example())
```

### Задание №3

Реализовать класс Stationery

```python
class Stationery:
    def __init__(self, title):
        self.title = title

    def draw(self):
        print('Запуск отрисовки')

class Pen(Stationery):
    def draw(self):
        print(f'{self.title}: ручка рисует тонкую ровную линию')

class Pencil(Stationery):
    def draw(self):
        print(f'{self.title}: карандаш рисует мягкий штрих')

class Handle(Stationery):
    def draw(self):
        print(f'{self.title}: маркер рисует широкую яркую линию')

print('\n=== Задание 3 ===')
for item in (Stationery('Общая'), Pen('Pen'), Pencil('Pencil'), Handle('Handle')):
    item.draw()
```

### Задание №4

```python
class BankAccount:
    bank_name = 'SberBank'
    accounts_count = 0

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        BankAccount.accounts_count += 1

    # методы экземпляра
    def deposit(self, amount):
        if not BankAccount.is_valid_amount(amount):
            print('Некорректная сумма')
            return
        self.balance += amount

    def withdraw(self, amount):
        if not BankAccount.is_valid_amount(amount) or amount > self.balance:
            print('Некорректная сумма или недостаточно средств')
            return
        self.balance -= amount

    # статический метод
    @staticmethod
    def is_valid_amount(amount):
        return isinstance(amount, (int, float)) and amount > 0

    # методы класса
    @classmethod
    def from_string(cls, data):
        owner, balance = data.split(';')
        return cls(owner.strip(), float(balance))

    @classmethod
    def total_accounts(cls):
        return cls.accounts_count

    def __str__(self):
        return f'{self.bank_name}: {self.owner}, баланс {self.balance:.2f}'

print('\n=== Задание 4 ===')
acc = BankAccount('Иван', 100)
acc.deposit(50)
acc.withdraw(30)
acc.withdraw(1000)
print(acc)

acc2 = BankAccount.from_string('Пётр; 500')
print(acc2)
print('Всего счетов:', BankAccount.total_accounts())
```
